## CPU reproduction

Run all cells with NumPy, pandas, SciPy and Matplotlib. Colab normally provides these; a clean local environment can use `requirements-v04-public-cpu.txt`. The notebook downloads the **public v0.4.0 input archive**, verifies its pinned digest and every internal file, and uses one shared evaluator. No model weights, credentials or RAMSES paths are needed.

**Scope:** reconstruct non-Botanic candidate ranks from cached scores; verify every model’s published rank and paired-outcome arithmetic.
```python
from pathlib import Path
import os, sys, hashlib, urllib.request, zipfile
ROOT = Path('public_inputs').resolve()
ARCHIVE = Path(os.environ.get('POPGENLM_PUBLIC_ARCHIVE', 'popgenlm-v04-public-inputs.zip'))
URL = 'https://github.com/tahirali-biomics/popgenlm-bench/releases/download/v0.4.0/popgenlm-v04-public-inputs.zip'
EXPECTED = '776c145febcc29d2c781a38c5a65b63fea2d5c647f0dd64d90e28d46f8676a9f'
if not ARCHIVE.exists():
    urllib.request.urlretrieve(URL, ARCHIVE)
assert hashlib.sha256(ARCHIVE.read_bytes()).hexdigest() == EXPECTED, 'Archive checksum mismatch'
if not ROOT.exists():
    with zipfile.ZipFile(ARCHIVE) as archive:
        for member in archive.infolist():
            assert ROOT in (ROOT / member.filename).resolve().parents, 'Unsafe archive path'
        archive.extractall(ROOT)
sys.path.insert(0, str(ROOT))
import review
manifest = review.verify(ROOT)
print('Verified', len(manifest['files']), 'files for', manifest['release'])
print(manifest['raw_reconstruction'])
```

```python
OUT = Path('outputs_plantcad')
counts = review.run(ROOT, OUT, 'plantcad')
print(counts.query("scope == 'model_covered'").to_string(index=False))
from IPython.display import display, Image
for figure in sorted(OUT.glob('*.png')):
    display(Image(filename=str(figure)))
```

```python
data = review.load(ROOT)
print(data['per_locus_outcomes'].query("scope == 'alpha_MLM_shared' and method == 'alphagenome_abs'")[["gene","baseline_rank","target_rank","baseline_hit","model_hit"]].to_string(index=False))
print('nemo B. napus transfer: 8/14 versus 8/14 on matched coverage; not native A. thaliana validation')
```
